```python
import pandas as pd

raw_df = pd.read_excel("messy_sales_orders.xlsx")
df = raw_df.copy()

df
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Exact duplicate rows:", df.duplicated().sum())

print("\nDuplicated order IDs:")
print(df[df.duplicated(subset=["order_id"], keep=False)])
```

```python
# PART 2 — CLEAN AND STANDARDIZE TEXT DATA

# 1. Clean leading/trailing spaces in text columns
text_columns = [
    "customer_name",
    "email",
    "phone",
    "country",
    "state",
    "city",
    "product_name",
    "category",
    "payment_method",
    "order_status"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()
# Standardize phone numbers
df["phone"] = (
    df["phone"]
    .str.replace(r"[()\s.-]", "", regex=True)
)


# 2. Standardize email casing
df["email"] = df["email"].str.lower()


# 3. Standardize states
state_mapping = {
    "California": "CA",
    "Texas": "TX",
    "Florida": "FL",
    "Colorado": "CO",
    "Ohio": "OH",
    "Missouri": "MO"
}

df["state"] = df["state"].replace(state_mapping)


# 4. Standardize payment methods
payment_mapping = {
    "credit card": "Credit Card",
    "Paypal": "PayPal"
}

df["payment_method"] = df["payment_method"].replace(payment_mapping)


# 5. Standardize order statuses
status_mapping = {
    "completed": "Completed",
    "COMPLETED": "Completed",
    "complete": "Completed",
    "IN PROGRESS": "Pending"
}

df["order_status"] = df["order_status"].replace(status_mapping)


# 6. Check the cleaned values
print("States:")
print(sorted(df["state"].dropna().unique()))

print("\nPayment methods:")
print(df["payment_method"].value_counts(dropna=False))

print("\nOrder statuses:")
print(df["order_status"].value_counts(dropna=False))
```

```python
# PART 3 — VALIDATE DATA

# 1. Validate emails
email_is_valid = (
    df["email"].notna()
    & df["email"].str.contains(
        r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$",
        regex=True,
        na=False
    )
)

# 2. Validate phone numbers
phone_is_valid = (
    df["phone"].notna()
    & df["phone"].str.match(r"^\d{10}$", na=False)
)

# 3. Validate ages
age_is_valid = (
    df["age"].notna()
    & df["age"].between(18, 100)
)

# 4. Validate quantity
quantity_is_valid = (
    df["quantity"].notna()
    & (df["quantity"] > 0)
)

# 5. Validate unit price
unit_price_is_valid = (
    df["unit_price"].notna()
    & (df["unit_price"] > 0)
)

# 6. Validate discount
discount_is_valid = (
    df["discount_pct"].notna()
    & df["discount_pct"].between(0, 100)
)

# 7. Detect duplicated order IDs
duplicate_order_id = df["order_id"].duplicated(keep=False)

# 8. Overall validation
df["is_valid"] = (
    email_is_valid
    & phone_is_valid
    & age_is_valid
    & quantity_is_valid
    & unit_price_is_valid
    & discount_is_valid
    & ~duplicate_order_id
)

# 9. Explain validation problems
def get_validation_issues(row):
    issues = []

    if pd.isna(row["email"]):
        issues.append("Missing email")
    elif not bool(
        pd.Series([row["email"]]).str.contains(
            r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$",
            regex=True
        ).iloc[0]
    ):
        issues.append("Invalid email")

    if pd.isna(row["phone"]):
        issues.append("Missing phone")
    elif not bool(
        pd.Series([row["phone"]]).str.match(r"^\d{10}$").iloc[0]
    ):
        issues.append("Invalid phone")

    if pd.isna(row["age"]):
        issues.append("Missing age")
    elif not 18 <= row["age"] <= 100:
        issues.append("Invalid age")

    if pd.isna(row["quantity"]) or row["quantity"] <= 0:
        issues.append("Invalid quantity")

    if pd.isna(row["unit_price"]) or row["unit_price"] <= 0:
        issues.append("Invalid unit price")

    if pd.isna(row["discount_pct"]) or not 0 <= row["discount_pct"] <= 100:
        issues.append("Invalid discount")

    if duplicate_order_id.loc[row.name]:
        issues.append("Duplicate order ID")

    return ", ".join(issues)


df["validation_issues"] = df.apply(get_validation_issues, axis=1)

# 10. Split valid and invalid records
clean_orders = df[df["is_valid"]].copy()
invalid_records = df[~df["is_valid"]].copy()

# 11. Check results
print("Valid records:", len(clean_orders))
print("Invalid records:", len(invalid_records))

print("\nInvalid records:")
print(
    invalid_records[
        [
            "order_id",
            "customer_name",
            "email",
            "phone",
            "age",
            "quantity",
            "unit_price",
            "discount_pct",
            "validation_issues"
        ]
    ]
)
```

```python
# PART 4 — CREATE BUSINESS VALUE

# 1. Create financial columns
clean_orders["gross_amount"] = (
    clean_orders["quantity"] * clean_orders["unit_price"]
)

clean_orders["discount_amount"] = (
    clean_orders["gross_amount"]
    * clean_orders["discount_pct"] / 100
)

clean_orders["net_amount"] = (
    clean_orders["gross_amount"]
    - clean_orders["discount_amount"]
)


# 2. Calculate main business statistics
total_revenue = clean_orders["net_amount"].sum()

average_order_value = clean_orders["net_amount"].mean()

total_orders = len(clean_orders)

completed_orders = (
    clean_orders["order_status"] == "Completed"
).sum()

pending_orders = (
    clean_orders["order_status"] == "Pending"
).sum()

cancelled_orders = (
    clean_orders["order_status"] == "Cancelled"
).sum()


# 3. Sales by product
sales_by_product = (
    clean_orders.groupby("product_name")["net_amount"]
    .sum()
    .sort_values(ascending=False)
)


# 4. Sales by state
sales_by_state = (
    clean_orders.groupby("state")["net_amount"]
    .sum()
    .sort_values(ascending=False)
)


# 5. Display results
print("Total revenue:", round(total_revenue, 2))
print("Average order value:", round(average_order_value, 2))
print("Total valid orders:", total_orders)
print("Completed orders:", completed_orders)
print("Pending orders:", pending_orders)
print("Cancelled orders:", cancelled_orders)

print("\nSales by product:")
print(sales_by_product)

print("\nSales by state:")
print(sales_by_state)
```

```python
# PART 5A — GENERATE FINAL EXCEL FILE

import pandas as pd

output_file = "clean_sales_report.xlsx"

# Create a summary table
summary_data = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Average Order Value",
        "Total Valid Orders",
        "Completed Orders",
        "Pending Orders",
        "Cancelled Orders"
    ],
    "Value": [
        round(total_revenue, 2),
        round(average_order_value, 2),
        total_orders,
        completed_orders,
        pending_orders,
        cancelled_orders
    ]
})

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    clean_orders.to_excel(
        writer,
        sheet_name="Clean Orders",
        index=False
    )

    invalid_records.to_excel(
        writer,
        sheet_name="Invalid Records",
        index=False
    )

    summary_data.to_excel(
        writer,
        sheet_name="Summary",
        index=False,
        startrow=0
    )

    sales_by_product.to_excel(
        writer,
        sheet_name="Summary",
        startrow=9,
        header=["Revenue"]
    )

    sales_by_state.to_excel(
        writer,
        sheet_name="Summary",
        startrow=20,
        header=["Revenue"]
    )

print(f"Excel file created: {output_file}")
```

```python
# PART 5B — FORMAT FINAL EXCEL FILE

from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Border, Side, Alignment
from openpyxl.utils import get_column_letter

workbook = load_workbook("clean_sales_report.xlsx")

# Basic styles
header_fill = PatternFill("solid", fgColor="1F4E78")
header_font = Font(color="FFFFFF", bold=True)

thin_border = Border(
    left=Side(style="thin"),
    right=Side(style="thin"),
    top=Side(style="thin"),
    bottom=Side(style="thin")
)

# Format all sheets
for sheet in workbook.worksheets:

    # Freeze top row
    sheet.freeze_panes = "A2"

    # Add filter to first row
    sheet.auto_filter.ref = sheet.dimensions

    # Style header row
    for cell in sheet[1]:
        cell.fill = header_fill
        cell.font = header_font
        cell.alignment = Alignment(horizontal="center")
        cell.border = thin_border

    # Add borders and auto width
    for column_cells in sheet.columns:
        max_length = 0
        column_letter = get_column_letter(column_cells[0].column)

        for cell in column_cells:
            cell.border = thin_border

            if cell.value is not None:
                cell_length = len(str(cell.value))
                if cell_length > max_length:
                    max_length = cell_length

        sheet.column_dimensions[column_letter].width = min(max_length + 2, 35)

# Format money columns in Clean Orders
clean_sheet = workbook["Clean Orders"]

for row in clean_sheet.iter_rows(min_row=2):
    for cell in row:
        if clean_sheet.cell(1, cell.column).value in [
            "unit_price",
            "gross_amount",
            "discount_amount",
            "net_amount"
        ]:
            cell.number_format = '$#,##0.00'

# Format Summary sheet
summary_sheet = workbook["Summary"]

for row in range(2, 8):
    metric = summary_sheet[f"A{row}"].value

    if metric in ["Total Revenue", "Average Order Value"]:
        summary_sheet[f"B{row}"].number_format = '$#,##0.00'

# Add section titles
summary_sheet["A9"] = "Sales by Product"
summary_sheet["A20"] = "Sales by State"

summary_sheet["A9"].font = Font(bold=True)
summary_sheet["A20"].font = Font(bold=True)

workbook.save("clean_sales_report.xlsx")

print("Excel formatting completed.")
```
